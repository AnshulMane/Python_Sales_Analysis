# Sales Analysis

#### Import necessary libraries

```python
import os
import pandas as pd
```

#### Merge data from each month into one CSV

```python
path = "./Sales_Data"
files = [file for file in os.listdir(path) if not file.startswith('.')] # Ignore hidden files

all_months_data = pd.DataFrame()

for file in files:
    current_data = pd.read_csv(path+"/"+file)
    all_months_data = pd.concat([all_months_data, current_data])
    
all_months_data.to_csv("all_data_copy.csv", index=False)
    
```

### Clean up the data!
The first step in this is figuring out what we need to clean. I have found in practice, that you find things you need to clean as you perform operations and get errors. Based on the error, you decide how you should go about cleaning the data

##### Drop rows of NAN

```python
# Find NAN
nan_df = all_data[all_data.isna().any(axis=1)]
display(nan_df.head())

all_data = all_data.dropna(how='all')
all_data.head()
```

##### Get rid of text in order date column

```python
all_data = all_data[all_data['Order Date'].str[0:2]!='Or']
```

#### Make columns correct type

```python
all_data['Quantity Ordered'] = pd.to_numeric(all_data['Quantity Ordered'])
all_data['Price Each'] = pd.to_numeric(all_data['Price Each'])
```

### Augment data with additional columns

#### Add month column

```python
all_data['Month'] = all_data['Order Date'].str[0:2]
all_data['Month'] = all_data['Month'].astype('int32')
all_data.head()
```

#### Add month column (alternative method)

```python
all_data['Month 2'] = pd.to_datetime(all_data['Order Date']).dt.month
all_data.head()
```

#### Add city column

```python
def get_city(address):
    return address.split(",")[1].strip(" ")

def get_state(address):
    return address.split(",")[2].split(" ")[1]

all_data['City'] = all_data['Purchase Address'].apply(lambda x: f"{get_city(x)}  ({get_state(x)})")
all_data.head()
```

#### Question 4: What products are most often sold together?

```python
# https://stackoverflow.com/questions/43348194/pandas-select-rows-if-id-appear-several-time
df = all_data[all_data['Order ID'].duplicated(keep=False)]

# Referenced: https://stackoverflow.com/questions/27298178/concatenate-strings-from-several-rows-using-pandas-groupby
df['Grouped'] = df.groupby('Order ID')['Product'].transform(lambda x: ','.join(x))
df2 = df[['Order ID', 'Grouped']].drop_duplicates()
```

```python
# Referenced: https://stackoverflow.com/questions/52195887/counting-unique-pairs-of-numbers-into-a-python-dictionary
from itertools import combinations
from collections import Counter

count = Counter()

for row in df2['Grouped']:
    row_list = row.split(',')
    count.update(Counter(combinations(row_list, 2)))

for key,value in count.most_common(10):
    print(key, value)
```

```python
# Referenced: https://stackoverflow.com/questions/14762181/adding-a-y-axis-label-to-secondary-y-axis-in-matplotlib

import matplotlib.pyplot as plt
import pandas as pd  # Assuming you're using pandas for data manipulation

# Assuming all_data has 'Product', 'Quantity Ordered', and 'Price Each' columns

# Calculate average price per product
prices = all_data.groupby('Product')['Price Each'].mean()

# Get product names (assuming it's a unique identifier)
product_names = prices.index.to_list()  # Extract product names from index

# Create the twin axes plot
fig, ax1 = plt.subplots()
ax2 = ax1.twinx()

# Plot quantity ordered (green bars)
ax1.bar(product_names, all_data.groupby('Product')['Quantity Ordered'].sum(), color='g')  # Use sum for total quantity

# Plot average price (blue line)
ax2.plot(product_names, prices, color='b')

# Set labels and tick rotations
ax1.set_xlabel('Product Name')
ax1.set_ylabel('Quantity Ordered', color='g')
ax2.set_ylabel('Average Price ($)', color='b')
ax1.set_xticklabels(product_names, rotation='vertical', size=8)

# Add a title (optional)
plt.title('Quantity Ordered vs. Average Price per Product')

plt.show()
```
